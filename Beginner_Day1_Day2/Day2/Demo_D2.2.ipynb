## Setup

Load the tools used by these examples.

```python
import numpy as np
import pandas as pd
```

## 1. Plot and inspect a sequence

**What and why.** Time order helps us see trend and repeated patterns.

**Look for.** Values generally rise. A longer series is needed to judge a repeating pattern.

```python
series = pd.Series([10,12,11,13,12,14,13,15])
print(series)
print('First:', series.iloc[0], 'Last:', series.iloc[-1])
```

## 2. Trend and repeated positions

**What and why.** A rolling mean shows local level. Grouping by position can reveal a repeating cycle.

**Look for.** The first two rolling values are missing because three observations are needed.

```python
series = pd.Series([10,12,11,13,12,14,13,15])
print('Rolling mean:', series.rolling(3).mean().round(1).tolist())
print('Position averages:', series.groupby(series.index % 2).mean().tolist())
```

## 3. Autocorrelation

**What and why.** We compare the series with its previous value.

**Look for.** The number describes the relationship with a one-step shift.

```python
series = pd.Series([10,12,11,13,12,14,13,15])
print('Lag 1 correlation:', round(series.autocorr(lag=1),2))
```

## 4. Lag and rolling average without leakage

**What and why.** We shift first so a feature only uses earlier values.

**Look for.** At the fourth row the previous-two mean is 11.5.

```python
series = pd.Series([10,12,11,13,12])
table = pd.DataFrame({'value': series})
table['lag1'] = series.shift(1)
table['past_mean'] = series.shift(1).rolling(2).mean()
print(table)
```

## 5. Ordered splits

**What and why.** Each training set ends before its test set begins.

**Look for.** Each test block comes after the corresponding training block.

```python
from sklearn.model_selection import TimeSeriesSplit
for train, test in TimeSeriesSplit(n_splits=3).split(np.arange(8)):
    print('Train:', train, 'Test:', test)
```
