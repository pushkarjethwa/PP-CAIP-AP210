# L2.3 · Model training and tuning

## What are we doing?

We will join inputs, compare model settings and look at errors. This shows why selection and testing require separate data.

Run one cell at a time with Shift+Enter. Read the explanation before running each cell.

## Setup

Load the tools used by these examples.

```python
import numpy as np
import pandas as pd
```

## 1. Join tables and inspect missing values

**What and why.** A join combines known attributes with features while making missing rows visible.

**Look for.** ID 3 has no activity value.

```python
left = pd.DataFrame({'id':[1,2,3], 'value':[10,20,30]})
right = pd.DataFrame({'id':[1,2], 'activity':[2,3]})
joined = left.merge(right,on='id',how='left')
print(joined); print('Missing:', joined['activity'].isna().sum())
```

## 2. Gradient descent updates

**What and why.** A learning rate controls how much the weight changes each step.

**Look for.** The value moves toward 3. Epochs count passes through training data and batch size counts examples per update.

```python
weight = 0.0
for number in range(3):
    gradient = 2 * (weight - 3)
    weight = weight - 0.25 * gradient
    print('Step', number + 1, ':', round(weight,2))
```

## 3. Baseline and regularised model

**What and why.** A baseline gives a simple reference for a learned line.

**Look for.** The baseline is 5; Ridge gives a prediction based on the input.

```python
from sklearn.linear_model import Ridge
x = [[1],[2],[3],[4]]
y = [2,4,6,8]
model = Ridge(alpha=1).fit(x,y)
print('Mean baseline:', np.mean(y))
print('Ridge prediction:', round(model.predict([[5]])[0],2))
```

## 4. Search for a setting

**What and why.** We compare several candidate values of alpha with cross-validation.

**Look for.** The grid checks three choices. Random search samples choices; halving spends more work on promising choices.

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
x = np.arange(1,9).reshape(-1,1)
y = np.arange(1,9)*2
search = GridSearchCV(Ridge(), {'alpha':[0.1,1.0,10.0]}, cv=2)
search.fit(x,y)
print('Chosen setting:', search.best_params_)
```

## 5. Nested cross-validation

**What and why.** An outer split provides a check while an inner process chooses settings.

**Look for.** Each outer test group stays outside that round's selection.

```python
from sklearn.model_selection import KFold
outer = KFold(n_splits=2)
for train, test in outer.split(np.arange(6)):
    print('Inner selection uses:', train, 'Outer check uses:', test)
```

## 6. Error and selection bias

**What and why.** We compare errors across observed groups and notice missing groups.

**Look for.** Group errors can differ. If a group is absent, we cannot measure its error here.

```python
check = pd.DataFrame({'group':['A','A','B'], 'actual':[10,20,30], 'predicted':[12,19,25]})
check['error'] = (check['actual'] - check['predicted']).abs()
print(check.groupby('group')['error'].mean())
```

## 7. Conservative estimates

**What and why.** A lower percentile gives a cautious value from a sample.

**Look for.** The sample 20th percentile is 11.6. A production estimate needs a validated method.

```python
estimates = pd.Series([10,12,14,16,18])
print('20th percentile:', estimates.quantile(0.2))
```
